Crear una funcion que grafique un circulo. Las variables de entrada seran las coordenadas del centro del circulo a,b el radio y el color del centro del circulo.

Llamar a la funcion al menos 5 veces.

No puede usarse la funcion circle ni patch

```python
import numpy as np
import matplotlib.pyplot as plt

def graficar_circulo(a, b, r, color):
    x = np.linspace(a - r, a + r, 1000) #Creamos los puntos de x (el dominio)

    #Aplicamos las fórmulas para las dos mitades
    y_superior = b + np.sqrt(r**2 - (x - a)**2)
    y_inferior = b - np.sqrt(r**2 - (x - a)**2)

    #Graficamos ambas partes con el mismo color
    plt.plot(x, y_superior, color=color)
    plt.plot(x, y_inferior, color=color)
    plt.gca().set_aspect('equal') # Esto asegura que el círculo no se vea como un óvalo

plt.figure(figsize=(6,6))

#Graficamos los 5 circulos
graficar_circulo(3, 5, 4, 'red')
graficar_circulo(4, 6, 2, 'green')
graficar_circulo(1, 7, 6, 'blue')
graficar_circulo(-1, -2, 4, 'grey')
graficar_circulo(0, 0, 6, 'yellow')

plt.title('Practica 2 Circulos')
plt.xlabel('Eje X')
plt.ylabel('Eje Y')
plt.gca().set_aspect('equal')
plt.grid(True)
plt.show()
```
