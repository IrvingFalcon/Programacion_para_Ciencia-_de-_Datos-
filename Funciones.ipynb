```python
#crear una funcion que regrese el factorial de un numero
def factorial(x):
  fac=1
  for i in range(x,1,-1):
    fac=fac*i
    return fac

def factorial2(x):
  if(x== 0 or x==1):
    return 1
  else:
    return x*factorial2(x-1)

#crear una funcion que devuelva un a lista con 10 numeros aleatorios

def lista():
  lista=[]
  for i in range(10):
    lista.append(np.random.rand())
  return lista

print(lista())

import matplotlib.pyplot as plt
#crear una funcion que tome como entrada un numero d y grafique una matriz
#aleatoria con dimension de d x d
def matriz(d):
  a=np.random.randint(-10,11,size=(d,d))
  plt.imshow(a)
```
