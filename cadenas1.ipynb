```python
# MANEJO DE CADENAS

# 1-Concatenacion combina 2 o mas cadenas en una
cadena1_concat = "hola"
cadena2_concat = "mundo"
resultado_concat = cadena1_concat + " " + cadena2_concat
print(f"Concatenación: '{resultado_concat}'")

# 2-Repeticion
cadena_rep = "python"
resultado_rep = cadena_rep * 3
print(f"Repetición de '{cadena_rep}': '{resultado_rep}'")

# 3-Longitud
cadena_len = "escuela"
longitud = len(cadena_len)
print(f"Longitud de '{cadena_len}': {longitud}")

# 4-Acceso a caracteres individuales
cadena_char = "ESCOM"
primer_caracter = cadena_char[0]
print(f"Primer caracter de '{cadena_char}': '{primer_caracter}'")

# 5-Rebanado
cadena_slice = "ciencia"
subcadena = cadena_slice[0:2]
print(f"Rebanado de '{cadena_slice}' [0:2]: '{subcadena}'")
```

```python
#Crear una funcion que reciba un numero entero y entregue una cadena de la longitud del numero
def cadena(numero):
    if not isinstance(numero, int) or numero < 0:
        raise ValueError("El numero debe ser un entero no negativo.")
    return 'i' * numero

# Ejemplo de uso:
longitud_deseada = 5
resultado_cadena = cadena(longitud_deseada)
print(f"La cadena con longitud {longitud_deseada} es: '{resultado_cadena}'")
print(f"La longitud de la cadena generada es: {len(resultado_cadena)}")
```

```python
def conteo(cadena, caracter):
    """
    Cuenta las ocurrencias de un caracter en una cadena.
    """
    return cadena.count(caracter)

# Ejemplo de uso:
mi_cadena = "banana"
letra_a = 'a'
conteo_a = conteo(mi_cadena, letra_a)
print(f"El caracter '{letra_a}' aparece {conteo_a} veces en '{mi_cadena}'.")

letra_n = 'n'
conteo_n = conteo(mi_cadena, letra_n)
print(f"El caracter '{letra_n}' aparece {conteo_n} veces en '{mi_cadena}'.")
```
