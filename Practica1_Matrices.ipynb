```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Definimos el tamaño de la matriz
n = 100

# 2. Creamos los gradientes base
# 'x' va de 0 (blanco) a 1 (negro) o viceversa
x = np.linspace(0, 1, n)
y = np.linspace(0, 1, n)
X, Y = np.meshgrid(x, y)

# 3. Generamos las 6 matrices de la imagen

A = -Y * 0.5 # Gradiente Vertical: Lo multiplicamos por 0.5 para que el blanco sea gris
B = -X * 0.5 # Gradiente Horizontal: Lo mismo, más oscuro
C = ((-Y - X) /2) ** 1 # Combinación Diagonal: Al elevarlo al cuadrado, las sombras crecen
D = (1 - np.sqrt((X-0.5)**2 + (Y-0.5)**2)) * 2 # Círculo Centro Blanco: Bajamos el brillo general
E = (np.sqrt((X-0.5)**2 + (Y-0.5)**2)) ** 2 # Círculo Centro Negro: Lo elevamos al cuadrado para expandir el negro
F = (np.abs(X-0.5) + np.abs(Y-0.5)) ** 4 # Diamante

# 4. Configuración de las gráficas
matrices = [A, B, C, D, E, F]
titulos = ['A', 'B', 'C', 'D', 'F', 'G'] # Etiquetas de cada imagen

plt.figure(figsize=( 8, 6))

for i in range(6):
    plt.subplot(2, 3, i + 1)
    #'gray' para blanco y negro
    plt.imshow(matrices[i], cmap='grey')
    plt.title(titulos[i])
    # Quitamos los ejes para que se vea igual a la proyección

plt.tight_layout()
plt.show()
```
