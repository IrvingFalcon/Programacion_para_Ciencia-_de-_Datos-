```python
import matplotlib.pyplot as plt

total_words = 0
frecuencia = {}

with open("/content/drive/MyDrive/nfl_texto_300_palabras.txt", encoding='utf8') as f:
  for line in f:#lee el texto linea por linea
   print(line.split())
   for palabras in line.split():#lee cada palabra del texto
    total_words+=1
    palabras  = palabras.lower()
    if palabras not in frecuencia:
      frecuencia[palabras] = 1
    else:
      frecuencia[palabras] += 1

print(frecuencia)
print("Total de palabras: ", total_words, "\n")

palabras_frecuentes = sorted(frecuencia, key=frecuencia.get, reverse=True)

print("Palabra \t Frecuencia")
print(palabras_frecuentes[:30])

for palabra in palabras_frecuentes[:30]:
  print(palabra, "\t", frecuencia[palabra])

y_frecuencia =[]
for palabra in palabras_frecuentes[:30]:
  y_frecuencia.append(frecuencia[palabra])

x_palabras = []
for palabra in palabras_frecuentes[:30]:
  x_palabras.append(palabra)


plt.scatter(frecuencia.keys(), frecuencia.values())
plt.show()

plt.xscale('log')
plt.yscale('log')
plt.bar(frecuencia.keys(), frecuencia.values())
plt.show()


```
