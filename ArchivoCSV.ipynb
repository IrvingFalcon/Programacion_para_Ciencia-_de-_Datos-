ARCHIVO CSV

Es un fomrato (valores separados por comas) utilizados para almacenar datos, como los quue se encunetran en una hoja
de calculo. Cada fila representa un registro y cada columna un atributo. Es compatible con excel y google sheets.
Para leer un archivo CSV utilizamos la libreria pandas que convierte un archivo CSV en un data frame.

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

fig, axes = plt.subplots(1, 1, figsize=(4, 4))
iris = pd.read_csv("https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv")
print((iris.shape))

pw = iris["petal_width"]
pl = iris["petal_length"]
sw = iris["sepal_width"]
sl = iris["sepal_length"]

print(len(pw))

A = np.zeros((150,4))
A[:,0] = pw
A[:,1] = pl
A[:,2] = sw
A[:,3] = sl

axes.scatter(pw, pl)
axes.set_ylabel("Longitud de petalo")
axes.set_xlabel("ancho de petalo")
plt.show()
```
