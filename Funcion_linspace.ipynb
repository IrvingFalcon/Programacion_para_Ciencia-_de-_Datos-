```python
import numpy as np
import matplotlib.pyplot as plt

plt.figure(figsize=(3,3))
x=np.linspace(0,10,100) #100 valores entre 0 y 10
y=0.5*x+1
plt.plot(x,y,color='r') #grafica x vs y
plt.ylabel('y')
plt.xlabel('x')
plt.ylim(0,10)# limite de la grafica y
plt.xlim(0,10)# limite de la grafica x
plt.grid()
```
