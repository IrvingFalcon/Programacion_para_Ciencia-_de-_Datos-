```python
import  numpy as np
import matplotlib.pyplot as plt

lista=[]
a=10
while a>0:
      lista.append(a)
      a+=-1
#print(lista)
lista=[]
a=10
while a>0:
      lista.append(a)
      a+=np.random.rand()-0.5

plt.plot(lista)
plt.xlabel('Iteraciones', size=20)
plt.ylabel('a', size=20)

```
