```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

fig, axes = plt.subplots(2, 3, figsize=(12, 8))#para definir tamaño y forma
iris = pd.read_csv("https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv")#Cargamos los datos que obtuvimos del archivo csv

colores = {'setosa':'red', 'versicolor':'orange', 'virginica':'green'}#asignamos colores a cada especie
#Lista de las combinaciones posibles
combinaciones = [
    ("petal_width", "petal_length", "ancho de pétalo", "longitud de pétalo"),
    ("petal_width", "sepal_width", "ancho de pétalo", "ancho de sépalo"),
    ("petal_width", "sepal_length", "ancho de pétalo", "longitud de sépalo"),
    ("petal_length", "sepal_width", "longitud de pétalo", "ancho de sépalo"),
    ("petal_length", "sepal_length", "longitud de pétalo", "longitud de sépalo"),
    ("sepal_width", "sepal_length", "ancho de sépalo", "longitud de sépalo")
    ]

axes = axes.flatten()

for i in range (6):#Generamos cada uno de los 6 gráficos
    ax=axes[i]
    col_x, col_y, etiqueta_x, etiqueta_y = combinaciones[i]

    for especie, color in colores.items():# Dibujamos los puntos por cada especie por separado para que la leyenda funcione bien
        datos_especie = iris[iris['species'] == especie]

        x = datos_especie[col_x]
        y = datos_especie[col_y]

# Creamos el gráfico de dispersión (scatter plot)
# alpha=0.8 añade transparencia y s=25 define el tamaño del punto
        ax.scatter(x, y, c=color, label=especie.capitalize(), alpha=0.8, s=25)
#etiquetas de 'x' y 'y'
    ax.set_xlabel(etiqueta_x)
    ax.set_ylabel(etiqueta_y)
    ax.legend(fontsize=8)

    ax.grid(True)#Añade una cuadrícula de fondo para mejor lectura

plt.tight_layout()
plt.show()
```
