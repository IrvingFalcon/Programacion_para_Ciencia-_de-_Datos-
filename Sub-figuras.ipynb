```python
import numpy as np
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1,3,figsize=(6,2)) #crea tres subfiguras en orden horizontal

a=np.ones((10,10)) #matriz de ls de 10 por 10
x=0
for i in range(0,10):#itera a i desde 0 hasta 9
  if x==0: #si x es igual a cero
    a[i,:]=0 #todos los valores de la fila i son iguales a 0
  x=1-x #cambia de 1 a 0 o 0 a 1 en cada iteracion
axes[1].imshow(a) #grafica la segunda subfigura

b=np.ones((10,10)) #matriz de ls de 10 por 10
x=0
for i in range(0,10):#itera a i desde 0 hasta 9
  if x==0: #si x es igual a cero
    b[:,i]=0 #todos los valores de la fila i son iguales a 0
  x=1-x #cambia de 1 a 0 o 0 a 1 en cada iteracion
axes[0].imshow(b) #grafica la primera subfigura

c=np.dot(a,b)
axes[2].imshow(c)
axes[2].set_title('axb')
plt.tight_layout()
plt.savefig('subfiguras.pdf', dpi=300)
```
